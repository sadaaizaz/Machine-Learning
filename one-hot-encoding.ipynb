# One Hot Encoding
**One-hot encoding is a data preprocessing technique used to convert categorical variables into a numerical format that machine learning algorithms can interpret. It replaces a single categorical column with multiple binary columns (0 or 1), representing the presence or absence of each category.**

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/kaggle/input/datasets/sadaaizaz/carsss/cars.csv')
```

# 1. OneHotEncoding using Pandas

```python
pd.get_dummies(df,columns=['fuel', 'owner'])
```

# 2. K-1 OneHotEncoding

```python
pd.get_dummies(df, columns=['fuel', 'owner'], drop_first=True)
```
